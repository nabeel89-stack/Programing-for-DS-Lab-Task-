## Dataset Loading

```python
import pandas as pd

df = pd.read_csv("train.csv")
print(df.head())
```

**Interpretation:** The dataset has 891 rows and 12 columns. Age has missing values, and the columns include both numerical and categorical types.

## Problem 2 - Data Filtering

```python
print(df[df["Pclass"] == 1])
```

```python
print(df[df["Age"] < 18])
```

```python
print(df[(df["Sex"] == "male") & (df["Survived"] == 1)])
```

```python
print(df[df["Fare"] > 100])
```

**Interpretation:** Filtering with Boolean conditions returns only the rows that match each condition.

## Problem 3 - Sorting and Aggregation

```python
print(df.sort_values(by="Fare", ascending=False).head(10))
```

```python
print(df["Age"].mean())
```

```python
print(df["Sex"].value_counts())
```

```python
print(df.groupby("Pclass")["Fare"].mean())
```

```python
print(df.groupby("Pclass")["Survived"].mean())
```

**Interpretation:** The average age is about 29.7. There are more male passengers than female passengers. Fare decreases and survival rate decreases from first class to third class.

## Findings

- The dataset contains 891 passengers with an average age of about 29.7 years.
- Third class has the most passengers.
- Female passengers paid a higher average fare than male passengers.
